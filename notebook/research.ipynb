##### Data Quality and Cleaning

```python
# How many ratings per user
ratings["user_id"].value_counts()
```

```python
#users with more than 200 ratings or read more than 200 books
x = ratings['user_id'].value_counts() > 200
```

```python
# Only 899 users have more than 200 ratings or read more than 200 books
x[x].shape
```

```python
#list if user_id that have more than 200 ratings or read more than 200 books
y = x[x].index
y
```

```python
#Extract only the ratings of users with more than 200 ratings (899) or read more than 200 books
ratings = ratings[ratings['user_id'].isin(y)]
```

```python
ratings.sample(200)
```

```python
# Now join ratings with books
ratings_with_books = ratings.merge(books, on='isbn')
```

```python
ratings_with_books.head(200)
```

```python
# Get the number of ratings per book
number_ratings = ratings_with_books.groupby('title')['rating'].count().reset_index()
number_ratings
```

```python
# Change the name of the column 'rating' to 'num_of_ratings'
number_ratings.rename(columns={'rating': 'num_of_ratings'}, inplace=True)
```

```python
# Add a new column to the "number_ratings" dataframe called 'num_ratings' and set it equal to the 'rating' column
# For each title repeat the number of ratings repeat
final_ratings = ratings_with_books.merge(number_ratings, on='title')
final_ratings.head()
```

```python
# Filter out books that have less than 50 ratings
final_ratings = final_ratings[final_ratings['num_of_ratings'] >= 50]
#Drop duplicates
final_ratings.drop_duplicates(['user_id', 'title'], inplace=True)
```

```python
#Check for null values
final_ratings.isnull().sum()
```

```python
# Final dataset with users that have more than 200 ratings and books that have more than 50 ratings
final_ratings.head()
```

##### Pivot Table - Preparing structure

user_id    |  224   225   226  
------------------------------------------
title      |  8.5   7.5   8.6
 abc       |  9.5   3.5   9.6
 def       |  5.5   6.5   5.0

user_id 224 y 225 seems to be similar (close vector)

```python
book_pivot = final_ratings.pivot_table(columns='user_id', index='title', values='rating')
```

```python
book_pivot.head()
```

```python
book_pivot.shape
```

```python
book_pivot.fillna(0, inplace = True)
```

```python
book_pivot.head()
```

##### Convert Pivot table TO CSR Matrix

<pre>
Too many Zeros, waste memory

              usuario1  usuario2  usuario3  usuario4
Libro A          9         0         0         0
Libro B          0         0         7         0
Libro C          0         8         0         0


Only data without zero

Libro A → usuario1: 9
Libro B → usuario3: 7
Libro C → usuario2: 8
</pre>

```python
from scipy.sparse import csr_matrix
```

```python
book_sparse_matrix = csr_matrix(book_pivot)
```

```python
book_sparse_matrix
```

```python
type(book_sparse_matrix)
```

##### Create the Model

```python
from sklearn.neighbors import NearestNeighbors
model = NearestNeighbors( algorithm='brute')
```

```python
model.fit(book_sparse_matrix)
```

```python
#One Row = All the ratings of the book "The Lovely Bones: A Novel" (2002) by Alice Sebold (237)
book_pivot.iloc[237,:]

```

```python
distances, suggestion = model.kneighbors(book_pivot.iloc[237,:].values.reshape(1,-1), n_neighbors=6)
```

```python
# Found the 6 nearest neighbors of the book "Harry Potter and the Chamber of Secrets (Book 2)"
distances
```

```python
# Index por books
suggestion
```

```python
# Books Recommended (position)
for i in range(len(suggestion)):
    print(book_pivot.index[suggestion[i]])
```

##### Testing model

```python
def recommend_books(book_name):
    book_id = np.where(book_pivot.index==book_name)[0][0]
    distances, suggestion = model.kneighbors(book_pivot.iloc[book_id,:].values.reshape(1,-1), n_neighbors=6)
    
    for i in range(len(suggestion)):
        books = book_pivot.index[suggestion[i]]
        for j in books:
            if j == book_name:
                print(f"You searched '{book_name}'\n")
                print("The suggestion books are: \n")
            else:
                print(j)
```

```python
book_name = "Harry Potter and the Chamber of Secrets (Book 2)"
recommend_books(book_name)
```
